```python
import plotly.graph_objects as go
import numpy as np

# Create sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig = go.Figure()

# Add main trace
fig.add_trace(go.Scatter(x=x, y=y, name='sine wave'))

# Add multiple shapes
fig.update_layout(
    shapes=[
        # Rectangle to highlight a region
        dict(
            type="rect",
            xref="x",
            yref="y",
            x0=2,
            y0=-0.5,
            x1=4,
            y1=0.5,
            fillcolor="LightSalmon",
            opacity=0.3,
            line=dict(width=0),
        ),
        # Vertical line
        dict(
            type="line",
            xref="x",
            yref="paper",
            x0=5,
            y0=0,
            x1=5,
            y1=1,
            line=dict(
                color="Green",
                width=2,
                dash="dash",
            ),
        ),
        # Circle
        dict(
            type="circle",
            xref="x",
            yref="y",
            x0=7,
            y0=-0.2,
            x1=8,
            y1=0.2,
            line=dict(color="RoyalBlue"),
            fillcolor="LightBlue",
            opacity=0.8,
        )
    ]
)

fig.show()
```
